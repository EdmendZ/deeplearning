```python
import torch
```

```python
tensor1 = torch.randint(low=0, high=10, size=(3, 5, 4))
print(tensor1)
```

```python
# 简单索引
print(tensor1[0, 1, 2])
print(tensor1[1])
print(tensor1[0, 1])
```

```python
# 取所有矩阵第二行
print(tensor1[:,1])
```

```python
# 范围索引
print(tensor1[1:])
```

```python
print(tensor1[-1, 1:4, 0:3])
```

```python
print(tensor1[-1, 1:4:2, 0:3:2])
# print(tensor1[-1, 4:1:-2, 0:3:2])   # 错误写法
```

```python
# 列表索引
print(tensor1)
```

```python
print(tensor1[[0, 2], [1, 2]])
```

```python
print(tensor1[[[0], [2]], [1, 2]])
```

```python
# 布尔索引
mask = tensor1 > 5
print(tensor1[mask])
```

```python
# 选取首元素大于5的所有行
mask = tensor1[:, :, 0] > 5
print(mask)
print(tensor1[mask])
```

```python
# 选取第二个元素大于5的所有列
mask = tensor1[:, 1, :] > 5
print(mask)
# print(tensor1[mask])
tensor2 = tensor1.permute(0, 2, 1)
print(tensor2)
print(tensor2[mask].T)
```

```python
mask = tensor1[:, 1, 3] > 5
print(tensor1[mask])
```

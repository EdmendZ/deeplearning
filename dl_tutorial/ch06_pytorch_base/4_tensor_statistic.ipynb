```python
import torch
```

```python
tensor1 = torch.randint(low=0, high=10, size=(3,2,4)).float()
print(tensor1)
```

```python
# sum求和
print("求和：", tensor1.sum())
```

```python
print("求和：", tensor1.sum(dim=2))
```

```python
# mean求均值
print("求均值:", tensor1.mean(dim=0))
```

```python
# std求标准差
print("求标准差：", tensor1.std(dim=1))
```

```python
# max/min
print("最大值：", tensor1.max())
```

```python
# argmax/argmin
print("最小值索引：", tensor1.argmin())
```

```python
# unique去重
print(tensor1.unique())
```

```python
tensor2 = torch.tensor([[[0., 8., 0., 7.],
         [6., 2., 6., 4.]],

        [[4., 3., 4., 0.],
         [7., 7., 7., 7.]],

        [[9., 0., 9., 2.],
         [3., 4., 3., 4.]]])
print(tensor2.unique(dim=2))
```

```python
# sort 排序
print(tensor1.sort(dim=0, descending=True))
```

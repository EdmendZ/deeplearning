```python
import torch
```

```python
x = torch.tensor([1.0, 2, 3], requires_grad=True)
y = x.detach()

print(x)
print(y)
```

```python
print(id(x))
print(id(y))
```

```python
print(x.untyped_storage().data_ptr())
print(y.untyped_storage().data_ptr())
```

```python
z1 = x ** 2
print(z1)
z2 = y ** 2
print(z2)
```

```python
print(z1.requires_grad)
z1.sum().backward()
```

```python
print(z2.requires_grad)
# z2.sum().backward()
```

```python
y[1] = 5
print(x)
print(y)
```

```python
import numpy as np
# 测试detach对梯度的影响
x = torch.tensor(np.array([[3., 1.], [2., 4.]]), requires_grad=True)
print(x)
y = x * x
print(y)
```

```python
# 用detach将y分离一份
u = y.detach()
print(u)
```

```python
# 用u和x进行计算
z = u * x
print(z)
```

```python
z.sum().backward()
```

```python
print(x.grad)
print(u.grad)
```

```python
print(x.grad == u)
```

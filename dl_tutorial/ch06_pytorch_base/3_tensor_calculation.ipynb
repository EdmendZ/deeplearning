```python
import numpy as np
import torch
```

```python
# 1. 基本运算
# 1.1 四则运算
tensor1 = torch.randint(low=1, high=10, size=(2,3))
print(tensor1)
print(tensor1 + 10)
print(tensor1)
```

```python
# 调用方法，非原地操作
print(tensor1.add(10))
print(tensor1)
```

```python
# 调用方法，原地操作
print(tensor1.add_(10))
print(tensor1)
```

```python
# 1.2 -、neg()、neg_()取负
print(-tensor1)
print(tensor1)
```

```python
print(tensor1.neg())
print(tensor1)
```

```python
print(tensor1.neg_())
print(tensor1)
```

```python
# 1.3 **、pow()、pow_()求幂
tensor1 = torch.tensor([1, 2, 3])
print(tensor1**2)
print(tensor1)
```

```python
print(tensor1.pow(2))
print(tensor1)
```

```python
print(tensor1.pow_(2))
print(tensor1)
```

```python
# 1.4 sqrt()、sqrt_()求平方根
print(tensor1.sqrt())
print(tensor1)
```

```python
# tensor2 = tensor1.type(torch.float)
tensor2 = torch.Tensor([1, 2, 3])
print(tensor2)
print(tensor2.sqrt_())
print(tensor2)
```

```python
# 1.5 exp()、exp_()以e为底数求幂
tensor1 = torch.tensor([1.0, 2, 3])
print(2.718281828459**tensor1)
print(tensor1.exp())
print(tensor1)
```

```python
print(tensor1.exp_())
print(tensor1)
```

```python
# 1.6 log()、log_()以e为底求对数
tensor1 = torch.tensor([1.0, 2, 3])
print(tensor1.log())
print(tensor1)
```

```python
print(tensor1.log_())
print(tensor1)
```

```python
# 2. 矩阵相关运算
# 2.1 哈达玛积
tensor1 = torch.tensor([[1, 2, 3], [4, 5, 6]])
tensor2 = torch.tensor([[1, 2, 3], [4, 5, 6]])
print(tensor1 + tensor2)
print(tensor1 * tensor2)
print(tensor1 ** 2)
print(tensor1.mul(tensor2))
```

```python
# 2.2 矩阵乘法
tensor1 = torch.tensor([[1, 2, 3], [4, 5, 6]])
tensor2 = torch.tensor([[1, 2],[3, 4], [5, 6]])
print(tensor1)
print(tensor2)
print(tensor1 @ tensor2)
print(tensor1.matmul(tensor2))
print(tensor1.mm(tensor2))
```

```python
# 张量的矩阵乘法
tensor1 = torch.tensor([[[1, 2, 3], [4, 5, 6]],  [[6, 5, 4], [3, 2, 1]]])  # 2×2×3
tensor2 = torch.tensor([ [[1, 2, 3, 4],[3, 4, 2, 1], [5, 6, 7, 8]],  [[6, 5, 4, 3], [3, 2, 2, 1], [1, 0, 5, 3]] ])  # 2×3×4
# tensor2 = torch.tensor([ [[1, 2],[3, 4], [5, 6]] ])  # 1×3×2
print(tensor1)
print(tensor2)
print(tensor1 @ tensor2)
print(tensor1.matmul(tensor2))
# print(tensor1.mm(tensor2))    # 错误
```

```python
# 内积：dot方法必须是一维向量相乘
tensor1 = torch.tensor([1, 2, 3])
tensor2 = torch.tensor([4, 5, 6])
print(tensor1.dot(tensor2))
```

```python
# 节省内存
X = torch.randint(low=1, high=10, size=(3,2,4))
Y = torch.randint(low=1, high=10, size=(3,2,4))
print(id(X))
X *= Y
print(id(X))
```

```python
X = torch.randint(low=1, high=10, size=(3,2,4))
Y = torch.randint(low=1, high=10, size=(3,4,4))
print(id(X))
print(X.shape)
# X = X @ Y
X @= Y
print(id(X))
print(X.shape)
```

```python
X = torch.randint(low=1, high=10, size=(3,2,4))
Y = torch.randint(low=1, high=10, size=(3,4,1))
# X = np.array([[1, 2, 3], [4, 5, 6]])
# Y = np.array([[1,2,3], [3, 4,5], [5, 6,7]])
print(id(X))
print(X.shape)
print(X)
print(Y)
# X[:] = X @ Y
X @= Y
print(id(X))
print(X)
```

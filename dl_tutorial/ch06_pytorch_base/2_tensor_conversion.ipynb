```python
import numpy as np
import torch
```

```python
# 1. 张量元素类型转换
tensor1 = torch.tensor([1, 2, 3])
print(tensor1, tensor1.dtype)
tensor1 = tensor1.type(torch.float64)
print(tensor1, tensor1.dtype)
```

```python
tensor1 = tensor1.int()
print(tensor1, tensor1.dtype)
```

```python
tensor1 = tensor1.long()
print(tensor1, tensor1.dtype)
```

```python
tensor1 = tensor1.bool()
print(tensor1, tensor1.dtype)
```

```python
tensor1 = tensor1.double()
print(tensor1, tensor1.dtype)
```

```python
np.set_printoptions(precision=8)
torch.set_printoptions(precision=8)
# 2. Tensor和ndarray的转换
# 2.1 将 Tensor 转为 ndarray
# 共享内存
tensor1 = torch.rand(3, 2)
np_array1 = tensor1.numpy()
print(tensor1, tensor1.dtype)
print(np_array1, np_array1.dtype)
print(type(tensor1), type(np_array1))
```

```python
tensor1[:, 0] = 10
print(tensor1)
print(np_array1)
```

```python
print(id(tensor1), id(np_array1))
```

```python
# 使用copy方法，避免共享内存
np_array2 = tensor1.numpy().copy()
tensor1[:, 0] = 100
print(tensor1)
print(np_array1)
print(np_array2)
```

```python
# 2.2 将ndarray转换为 Tensor
# 共享内存
np_array = np.random.randn(3, 2)
tensor1 = torch.from_numpy(np_array)
print(tensor1, tensor1.dtype)
print(np_array)
```

```python
np_array[0, 1] = 5
print(np_array)
print(tensor1)
```

```python
# 使用copy方法，避免共享内存
tensor2 = torch.from_numpy(np_array.copy())
np_array[0, 1] = 8
print(np_array)
print(tensor1)
print(tensor2)
```

```python
# 2.3 使用 tensor 方法，将ndarray转换成 Tensor；不共享内存
np_array = np.random.randn(3, 2)
tensor1 = torch.tensor(np_array)
print(np_array)
print(tensor1)
```

```python
np_array[0, 1] = 5
print(np_array)
print(tensor1)
```

```python
# 3. 将单一元素的张量转换成标量
tensor1 = torch.tensor([[[2]]])
print(tensor1)
print(tensor1.item())
```

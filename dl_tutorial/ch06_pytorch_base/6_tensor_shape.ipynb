```python
import torch
```

```python
# 交换维度
tensor1 = torch.randint(low=0, high=10, size=(3, 2, 5))
print(tensor1)
```

```python
print(tensor1.transpose(-1, -2))
```

```python
print(tensor1.transpose(0, 2))
```

```python
print(tensor1.permute(1, 2, 0))
```

```python
# 调整形状
print(tensor1.reshape(5, 6))
print(tensor1.reshape(3, -1))
```

```python
# view视图
print(tensor1.is_contiguous())
```

```python
print(tensor1.view(-1, 10))
```

```python
tensor1 = tensor1.T
print(tensor1)
print(tensor1.is_contiguous())
```

```python
print(tensor1.contiguous().view(-1))
```

```python
# 增加维度
tensor1 = torch.tensor([1, 2, 3])
print(tensor1.shape)
print(tensor1.unsqueeze(dim=0))
```

```python
print(tensor1.unsqueeze(dim=1))
```

```python
tensor2 = torch.tensor([[1, 2, 3], [4, 5, 6]])
print(tensor2.shape)
print(tensor2.unsqueeze(dim=2))
```

```python
# 删除大小为1的维度
# 先增加维度1
print(tensor2.unsqueeze_(dim=1))
print(tensor2)
```

```python
print(tensor2.shape)
# 再删除维度1
print(tensor2.squeeze_(dim=1))
print(tensor2.shape)
```

```python
# 张量拼接
tensor1 = torch.randint(low=0, high=10, size=(2, 2, 5))
tensor2 = torch.randint(low=0, high=10, size=(2, 1, 5))
print(tensor1)
print(tensor2)
```

```python
print(torch.cat([tensor1, tensor2], dim=1))
```

```python
tensor1 = torch.randint(low=0, high=10, size=(3, 1, 5))
tensor2 = torch.randint(low=0, high=10, size=(3, 1, 5))
print(tensor1)
print(tensor2)
```

```python
# 张量堆叠
tensor3 = torch.stack([tensor1, tensor2])
print(tensor3)
print(tensor3.shape)
```

```python
tensor3 = torch.stack([tensor1, tensor2], dim=3)
print(tensor3)
print(tensor3.shape)
```

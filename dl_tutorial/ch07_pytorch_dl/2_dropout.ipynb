```python
import torch
```

```python
# 随机生成一组x
x = torch.randint(low=1, high=10, size=(10,), dtype=torch.float)
print(x)
```

```python
# 定义一个Dropout层
dropout = torch.nn.Dropout(p=0.2)
```

```python
print(dropout(x))
```

```python
x = torch.randn(10, 3, device=torch.device('cuda'))
print(x.device)
```

```python
x2 = torch.randn(10, 3).to('cuda')
print(x2.device)
```

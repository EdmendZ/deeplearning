```python
import torch
import torch.nn as nn
```

## 全连接层

```python
linear = nn.Linear(in_features=2, out_features=3, bias=True)
```

```python
print(linear.weight)
print(linear.bias)
```

## 向前传播

```python
# 单条数据
#x = torch.tensor([1, 2]).float()

x = torch.randn(10, 2)

y = linear.forward(x)

print(y)
```

```python
y = linear(x)  ## 相当于调用: linear.__call__(). 每个module都会在__call__内调用 self.forward
print(y)
```

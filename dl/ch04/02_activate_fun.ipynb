```python
import torch
import torch.nn as nn
```

```python
x = torch.tensor([0.7, -1.2, 2.5, 0, -3.1])
```

## Sigmoid函数

```python
## 方式1: 调用torch的sigmoid函数
y = torch.sigmoid(x)
print(y)
```

```python
## 方式2: 直接调用tensor的sigmoid方法
y = x.sigmoid()
print(y)
```

```python
## 方式3: pytorch把所有的激活函数也当成类实现
sigmoid = nn.Sigmoid()
#y = sigmoid.forward(x)
y = sigmoid(x)
print(y)
```

## tanh() 双曲正切函数

```python
print(torch.tanh(x))
print(x.tanh())
print(nn.Tanh()(x))
```

## ReLu函数

```python
print(torch.relu(x))
print(x.relu())
print(nn.ReLU()(x))
```

## softmax 函数

```python
print(torch.softmax(x, dim=-1))
print(x.softmax(dim=-1))
print(nn.Softmax(dim=-1)(x))
```

## 其他激活函数

```python
print(nn.LeakyReLU()(x))
print(nn.PReLU()(x))
print(nn.ReLU()(x))
print(nn.ELU()(x))
print(nn.Softplus()(x))
print(nn.GELU()(x))
```

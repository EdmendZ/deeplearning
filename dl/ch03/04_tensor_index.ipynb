```python
import torch
t = torch.randint(0, 10, (2, 5, 4))
print(t)
print(t.shape)
```

## 简单索引

```python
print(t[0][0][0])
print(t[1][1])

print(t[0,0,0])
print(t[1,1])
```

## 范围索引

```python
print(t[1, 1:])
```

```python
print(t[1, 1:-2:2])
```

## 列表索引

```python
list1=[0,1]
list2=[1,2]
list3=[2,2]
print(t[list1,list2, list3])
```

```python
list1=[0,1]
list2=[1,2]
print(t[list1,list2])
```

```python
list1=[[0],[1]]
list2=[1,2]
print(t[list1,list2])
```

```python
list1 = [0,1,3,4,5,2]
# 变成独热编码
t1 = torch.zeros(6, 6)
t1[[0,1,2,3,4,5], list1] = 1
print(t1)
```

## 布尔索引

```python
# 取出所有的大于5的值
mask = t > 5
print(mask)
```

```python
print(t[mask])
```

```python
# 取出所有行中,首元素于5的那些行
mask = t[:, :, 0] > 5
print(mask)
```

```python
print(t[mask])
```

```python
# 取出所有列中,首元素于5的那些列
mask = t[:, 0, :] > 5
print(mask)
```

```python
print(t.mT[mask].mT)
```

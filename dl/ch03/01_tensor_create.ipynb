```python
import torch
from numpy.lib.tests.test_format import dtype
```

## 1. 基于内容创建tensor

```python
# 1. 标量
t1 = torch.tensor(10)
print(t1)
print(t1.shape)
```

```python
# 2. 列表
t1 = torch.tensor([10, 20, 30])
print(t1)
print(t1.shape)
```

```python
# 3. 矩阵
t1 = torch.tensor([[10, 20, 30], [100, 200, 300]])
print(t1)
print(t1.shape)
```

```python
# 4. 矩阵
t1 = torch.tensor([ [[10, 20, 30], [100, 200, 300]] , [[10, 20, 30], [100, 200, 300]]])
print(t1)
print(t1.shape)
```

```python
# 5. numpy的ndarray
import numpy as np
t1 = torch.tensor(np.array([[10, 20, 30], [100, 200, 300]]))
print(t1)
print(t1.shape)
```

```python
## 数据类型: 默认是根据传入数据的类型
t1 = torch.tensor([[10, 20, 30], [100, 200, 300]])
print(t1)
print(t1.shape)
print(t1.dtype)
```

## 2. 基于形状创建tensor

```python
t1 = torch.Tensor(2,3)  # 类型默认float32
print(t1)
print(t1.shape)
print(t1.dtype)
```

## 总结
1. 基于内存创建用小写tensor, 类型由传入的数据类定
2. 基于形状创建用的大写的tensor, 类型是默认float32

## 3. 创建指定类型的tensor

```python
t1 = torch.tensor([1,3], dtype=torch.float)
print(t1)
print(t1.shape)
print(t1.dtype)
```

```python
t1 = torch.tensor([1,3]).float()
print(t1)
print(t1.shape)
print(t1.dtype)
```

```python
t1 = torch.tensor([1,3]).to(torch.float)
print(t1)
print(t1.shape)
print(t1.dtype)
```

```python
t1 = torch.LongTensor(2,3)  # int64
t1 = torch.IntTensor(2,3) # int32
t1 = torch.ShortTensor(2,3) # int16
t1 = torch.ByteTensor(2,3) # uint8
t1 = torch.FloatTensor(2,3) # float32
t1 = torch.DoubleTensor(2,3) # float64
t1 = torch.HalfTensor(2,3) # float16
t1 = torch.BoolTensor(2,3) # Bool
print(t1)
print(t1.shape)
print(t1.dtype)
```

## 4. 基于区间创建

```python
t1 = torch.arange(start=1, end=10, step=2)  # [1, 10)
print(t1)
print(t1.shape)
print(t1.dtype)
```

```python
t1 = torch.arange(10)  # [0, 10)
print(t1)
print(t1.shape)
print(t1.dtype)
```

```python
t1 = torch.linspace(1, 9, 10)  # [start, end] step = (end-start)/(steps - 1)
print(t1)
print(t1.shape)
print(t1.dtype)
```

```python
t1 = torch.logspace(1, 3, 5, base=torch.e)  #  10^1  10^2 10^3
print(t1)
print(t1.shape)
print(t1.dtype)
```

## 5. 基于填充创建

```python
t1 = torch.ones(3, 4)
print(t1)
print(t1.shape)

t2 = torch.ones_like(t1)
print(t2)
print(t2.shape)
```

```python
t1 = torch.zeros(3, 4)
print(t1)
print(t1.shape)

t2 = torch.zeros_like(t1)
print(t2)
print(t2.shape)
```

```python
t2 = torch.empty(2,3)
print(t2)
print(t2.shape)
```

```python
t2 = torch.full((2,3), 10)
print(t2)
print(t2.shape)
```

```python
t2 = torch.eye(5)
print(t2)
print(t2.shape)
```

## 6. 随机创建

```python
## 均匀分布: [0, 1)
t1 = torch.rand(2,3)
print(t1)
print(t1.shape)
```

```python
## 均匀分布整数
t1 = torch.randint(1, 10, (2,3))
print(t1)
print(t1.shape)
```

```python
## 正态分布: 标准正态分布: 均值0 标准差1
t1 = torch.randn(2,3)
print(t1)
print(t1.shape)
```

```python
## 正态分布: 标准正态分布: 均值0 标准差1
t1 = torch.normal(mean=0, std=1, size=(2,3))
print(t1)
print(t1.shape)
```

```python
# 洗牌
t1 = torch.randperm(100)
print(t1)
print(t1.shape)
```

```python
# 查看随机数种子
print(torch.initial_seed())
```

```python
import torch
```

## 类型的转换

```python
t1 = torch.tensor([10, 20])
```

```python
t2 = t1.type(torch.float32)
print(t2)
print(t2.dtype)
```

```python
t2 = t1.to(torch.float32)
print(t2.dtype)
```

```python
t2 = t1.float()
print(t2.dtype)
```

## tensor=>ndarray的转换

```python
## tensor => ndarray
t1 = torch.tensor([10, 20])
n1 = t1.numpy() # 内存共享
print(type(n1))
print(n1)
```

```python
t1[0] = 100
print(t1)
print(n1)
```

```python
#避免内存共享
n2 = t1.numpy().copy()
print(n2)
t1[0] = 200
print(n2)
```

## ndarray=>tensor

```python
import numpy as np

n1 = np.array([10, 20])
```

```python
t1 = torch.from_numpy(n1)  # 内存共享
print(t1)
n1[0] = 100
print(t1)
```

```python
t1 = torch.from_numpy(n1).clone()
print(t1)
n1[0] = 200
print(t1)
```

## tensor=>标量

```python
t1 = torch.tensor([10, 20])
```

```python
print(t1[0].item())  # tensor的内容只有一个元素
```

```python
t1 = torch.tensor([[[10]]])
print(t1.shape)
print(type(t1.shape))
print(t1.item())
```

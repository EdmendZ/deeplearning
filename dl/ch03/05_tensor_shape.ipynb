```python
import torch
t1 = torch.randint(1, 10, (3,2,6))
print(t1)
print(t1.shape)
```

## 维度变换

```python
## 转置的本质: 索引的交换   (1,1,4) => (4,1,1)
print(t1.T)
print(t1.T.shape)
```

```python
print(t1.mT)
```

```python
## 交换任意两个维度
print(t1.transpose(0, 1))  # (1,0,0) => (0,1,0)
```

```python
# 重排所有维度
t2 = t1.permute(2,1,0)
print(t2)
print(t2.shape)
```

## shape的调制

```python
# 1. reshape
print(t1.reshape(-1))
```

```python
print(t1.reshape(-1,9))
```

```python
print(t1.reshape(3,1, 2,-1))
```

```python
# view
print(t1.view(3, 1, 2, -1))
```

```python
print(t1.mT.reshape(-1))
```

```python
#print(t1.mT.view(-1))
print(t1.is_contiguous())
print(t1.mT.is_contiguous())

# 强制内存连续
print(t1.mT.contiguous().view(-1))
```

## 增删维度(增加维度1或者删除所有的1的维度)

```python
t2 = t1.unsqueeze(dim=-1)
print(t2.shape)
print(t2)
```

```python
t2 = t1.reshape(1,1,1,-1)
print(t2.shape)
t3 = t2.squeeze()
print(t3.shape)
print(t3)
```

## tensor的拼接

```python
t1 = torch.randint(1, 10, (2, 1, 3))
t2 = torch.randint(1, 10, (2, 3, 3))
print(t1)
print(t2)
```

```python
t3 = torch.cat((t1, t2), dim=1)  # 处理dim=1 , 其他必须匹配
print(t3)
```

```python
# stack: 先做 unsqueeze, 然后再 cat
t1 = torch.randint(1, 10, (2, 3, 3))
t2 = torch.randint(1, 10, (2, 3, 3))

t3 = torch.stack((t1, t2), dim=0)
print(t3)
print(t3.shape)

```

```python
t1 = torch.randint(1, 10, (2, 3))
t2 = torch.randint(1, 10, (2, 3))
t3 = torch.vstack((t1,t2))
print(t1)
print(t2)
print(t3)
```

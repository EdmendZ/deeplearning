```python
import torch
```

## 1. 基本运算 + - * /

```python
t1 = torch.tensor([10, 20, 30])
print(t1 + 10)

print(t1.add(10))

print(t1)
t1.add_(10)  # 修改t1的内容
print(t1)
```

## 元素级运算 (tensor和tensor)

```python
t1 = torch.tensor([[1, 2, 3], [4, 5, 6]])
t2 = torch.tensor([[10, 20, 30], [40, 50, 60]])
```

```python
# tensor+tensor
t2 = torch.tensor([[10], [60]])  # (2,3) (2, 1) 可以广播
print(t1 + t2)
```

```python
# tensor+tensor  对位相乘  哈达玛积
print(t1 * t2)
print(t1.mul(t2))
print(t1.mul_(t2))
print(t1)
```

## 矩阵相乘


```python
t1 = torch.tensor([[1, 2, 3], [4, 5, 6]])
t2 = torch.tensor([[10, 20, 30], [40, 50, 60], [70, 80, 90]])
```

```python
print(t1 @ t2)
print(t1.mm(t2))
print(t1.matmul(t2))
```

```python
t1 = torch.tensor([[[1, 2, 3], [4, 5, 6], [6, 5, 4]], [[3, 2, 1], [3, 2, 1], [3, 2, 1]]])
t2 = torch.tensor([[[1, 2], [3, 4], [5, 6]], [[6, 5], [4, 3], [2, 1]]])
```

```python
print(t1.shape, t2.shape)
print(t1 @ t2)  # 高维度相乘: 最后两个维度满足矩阵相乘, 其他维度保持一致或者满足广播原则
print(t1.matmul(t2))
print(t1.mm(t2))  # mm只能用于矩阵
```

## 统计运算

```python
t1 = torch.randint(1, 10, (2, 3, 5))
print(t1)
print(t1.shape)
```

```python
print(t1.sum())
```

```python
print(t1.sum(dim=0))
```

```python
print(t1.sum(dim=1))
```

```python
print(t1.sum(dim=2))
```

```python
print(t1.max())
print(t1.argmax())  # 最大值的下标
```

```python
print(t1.max(dim=1))
```

```python
## 均值
print(t1.float().mean())
```

```python
print(t1.float().mean(dim=0))
```

```python
# var和std
print(t1.float().std())
print(t1.float().var())
```

```python
print(t1.float().std(dim = 1))
print(t1.float().var(dim = 1))
```

```python
t1[1,0,0] = 5
print(t1.float().unique())
```

```python
print(t1.float().sort(dim=0)) # 默认=>dim=2
```

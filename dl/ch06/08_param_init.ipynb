```python
import torch
import torch.nn as nn

# 创建一个全连接层(输入2维, 输出3维), 其 weight 形状为 (3, 2)
# 下面演示 torch.nn.init 提供的各种参数初始化方法
linear = nn.Linear(2, 3)
```

## 常数初始化

```python
print(linear.weight)
# 常数初始化: 把权重全部置为同一个值
#nn.init.zeros_(linear.weight)   # 全部置 0
nn.init.ones_(linear.weight)     # 全部置 1
nn.init.constant_(linear.weight, 5)  # 全部置为指定常数 5
print(linear.weight)
```

## 秩初始化

```python
# 单位矩阵初始化: 主对角线为 1, 其余为 0(非方阵时尽量填充对角)
nn.init.eye_(linear.weight)
print(linear.weight)
```

## 正态分布初始化

```python
# 正态(高斯)分布初始化: 从均值 mean、标准差 std 的正态分布中采样
nn.init.normal_(linear.weight, mean=0, std=1)
print(linear.weight)
```

## 均匀分布初始化

```python
# 均匀分布初始化: 从区间 [a, b] 上的均匀分布中采样
nn.init.uniform_(linear.weight, a=1, b=2)
print(linear.weight)
```

## Xavier 初始化

```python
# Xavier 初始化: 根据输入和输出维度自动设定方差, 适合 Sigmoid/Tanh 等激活函数
## 均匀分布版本
nn.init.xavier_uniform_(linear.weight)
print(linear.weight)
```

```python
## 正态分布版本
nn.init.xavier_normal_(linear.weight)
print(linear.weight)
```

## He初始化(Kaiming初始化)

```python
# He(Kaiming) 初始化: 专为 ReLU 系列激活函数设计, mode 指定按扇入(fan_in)还是扇出(fan_out)计算方差
## 均匀分布版本
nn.init.kaiming_uniform_(linear.weight, mode='fan_in', nonlinearity='relu')
print(linear.weight)

## 正态分布版本
nn.init.kaiming_normal_(linear.weight, mode='fan_in', nonlinearity='relu')
print(linear.weight)
```

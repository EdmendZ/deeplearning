```python
import torch
import torch.nn as nn
from torch.optim import SGD

# 本节演示三种常用的正则化 / 训练稳定化手段:
# 批归一化(BatchNorm)、权值衰减(weight_decay)、随机失活(Dropout)
```

## BN层

```python
# 构造一批形状为 (5, 3) 的样本: 5 个样本, 每个 3 个特征
x = torch.randint(0, 10, (5, 3)).float()
print(x)
```

```python
# 创建批归一化层, num_features 对应特征数(列数), 会按列(每个特征)做归一化
bn = nn.BatchNorm1d(num_features=3)
```

```python
# 对每个特征列减均值除标准差, 使输出近似零均值、单位方差
y = bn(x)
print(y)
```

## 权值衰减

```python
# 权值衰减(L2 正则): 在优化器中设置 weight_decay, 会在梯度上额外加上权重惩罚项,
# 抑制权重过大从而缓解过拟合。(此处仅演示参数写法, 并非可直接运行的完整调用)
SGD(weight_decay=0.2)
```

## Dropout

```python
# 构造一个长度为 10 的一维张量, 用于演示 Dropout 的随机失活效果
x = torch.randint(1, 10, (10, )).float()
print(x)
```

```python
# Dropout: 训练时以概率 p 将元素随机置 0, 其余元素放大 1/(1-p) 倍以保持期望不变
dp = nn.Dropout(p = 0.5)
y = dp(x)
print(y)
```

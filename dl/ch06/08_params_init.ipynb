```python
import torch
import torch.nn as nn
```

```python
# 定义一个全连接处
linear = nn.Linear(2, 3)
```

## 常数初始化

```python
print(linear.weight)
print(linear.bias)
nn.init.zeros_(linear.weight)
nn.init.zeros_(linear.bias)
print(linear.weight)
print(linear.bias)
```

```python
nn.init.ones_(linear.weight)
nn.init.ones_(linear.bias)
print(linear.weight)
print(linear.bias)
```

```python
nn.init.constant_(linear.weight, 10)
nn.init.constant_(linear.bias, 10)
print(linear.weight)
print(linear.bias)
```

```python
## 特殊常数初始化(秩初始化)
nn.init.eye_(linear.weight)
print(linear.weight)
```

## 随机初始化

```python
# 把参数初始化位正态分布: 默认是标准正态分布
nn.init.normal_(linear.weight, mean=0, std=1)
print(linear.weight)
```

```python
# 把参数初始化均匀分布
nn.init.uniform_(linear.weight, a=1, b=2)
print(linear.weight)
```

## xavier 初始化

```python
nn.init.xavier_normal_(linear.weight)
```

```python
nn.init.xavier_uniform_(linear.weight)
```

## He初始化(凯明初始化)

```python
nn.init.kaiming_normal_(linear.weight)
print(linear.weight)
```

```python
nn.init.kaiming_uniform_(linear.weight)
print(linear.weight)
```

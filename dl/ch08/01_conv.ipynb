```python
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
```

```python
# 1. 读取图片
img = plt.imread("../data/duck.jpg")
print(img.shape)
```

```python
# 2. 张量的转换
img_data = torch.tensor(img).permute(2, 0, 1).float()
print(img_data.shape)
```

```python
# 3. 定义卷积层
conv = nn.Conv2d(3, 3, kernel_size=3, padding=1, stride=2)

cov_img = conv(img_data)

print(cov_img.shape)
```

```python
# 转换图片想着
cov_img = torch.clamp(cov_img, min=0, max=255).to(torch.uint8)
out_img = cov_img.permute(1, 2, 0)
```

```python
fig, ax = plt.subplots(1, 2, figsize=(10, 5))
ax[0].imshow(img)

ax[1].imshow(out_img.detach())

plt.show()
```

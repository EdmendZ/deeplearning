```python
import torch
import torch.nn as nn
```

## (BCE)二元交叉熵损失函数

```python
# 1. 定义条数据, 模拟神经网络的输出
output = torch.randn(5)

# 2. 使用sigmoid映射成概率(1的概率)
proba = torch.sigmoid(output)

# 5. 定义正确的目标标签
target = torch.tensor([0, 1, 1, 0, 1]).float()
```

```python
loss_fn = nn.BCELoss()
loss = loss_fn(proba, target)
print(loss)
```

```python

# 把BCE和sigmoid整合到了一起
loss_fn = nn.BCEWithLogitsLoss()
loss = loss_fn(output, target)
print(loss)
```

## 交叉熵损失函数

```python
# 模拟神经网络的输出, 10条数据, 6个分类的数据
output = torch.randn(10, 6)
# 目标值: 标签编码(顺序)
print(target)
loss_fn = nn.CrossEntropyLoss()
loss = loss_fn(output, target)

loss_fn = nn.CrossEntropyLoss()
loss = loss_fn(output, target)
print(loss)
```

```python
# 假设目标是独热编码
target_one = torch.zeros(10, 6)
target_one[torch.arange(10), target] = 1

loss_fn = nn.CrossEntropyLoss()
# softmax + NLLoss
loss = loss_fn(output, target_one)
print(loss)
```

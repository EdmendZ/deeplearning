```python
import torch
import torch.nn as nn
```

## (BCE)二元交叉熵损失函数

```python
# 1. 定义一条数据, 模拟神经网络的输出（原始分数 logits）
output = torch.randn(5)

# 2. 使用sigmoid映射成概率（属于正类/1 的概率），范围 (0,1)
proba = torch.sigmoid(output)

# 3. 定义正确的目标标签（二分类，0 或 1），BCE 要求 target 为 float
target = torch.tensor([0, 1, 1, 0, 1]).float()
```

```python
# BCELoss：输入必须是「概率」（已过 sigmoid），不能直接传 logits
loss_fn = nn.BCELoss()
loss = loss_fn(proba, target)
print(loss)
```

```python
# BCEWithLogitsLoss：把 sigmoid 和 BCE 整合到了一起，直接吃原始 logits
# 数值更稳定，实战推荐用这个（模型最后一层就不用再加 sigmoid），结果与上面一致
loss_fn = nn.BCEWithLogitsLoss()
loss = loss_fn(output, target)
print(loss)
```

## 交叉熵损失函数

```python
# 模拟神经网络的输出: 10条数据, 6个分类的原始分数（logits）
output = torch.randn(10, 6)
# 目标值: 标签编码（每个元素是类别下标，shape=(10,)，整型）
target = torch.tensor([1, 2, 1, 4, 5, 3, 0, 3, 4, 3])
print(target)
# CrossEntropyLoss 内部 = LogSoftmax + NLLLoss，自带 softmax
# 所以输入是 logits（不要再加 softmax），target 传类别下标即可
loss_fn = nn.CrossEntropyLoss()
loss = loss_fn(output, target)
print(loss)
```

```python
# CrossEntropyLoss 的 target 也支持独热编码（one-hot，shape=(10,6)，浮点）
# 把标签编码转成独热：每行在对应类别下标处置 1
target_one = torch.zeros(10, 6)
target_one[torch.arange(10), target] = 1

# 用独热编码算出的 loss 与上面标签编码版完全相同
loss_fn = nn.CrossEntropyLoss()
loss = loss_fn(output, target_one)
print(loss)
```

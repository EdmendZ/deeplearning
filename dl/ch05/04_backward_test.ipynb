```python
import torch
import torch.nn as nn
```

```python
# 1. 准备数据
x = torch.tensor([10.0], requires_grad=True)
y = torch.tensor([20.0], requires_grad=True)
```

```python
# 2. 定义参数
w = torch.tensor(1.0, requires_grad=True)
b = torch.tensor(1.0, requires_grad=True)
print(w, b)
```

```python
# 3. 向前传播
z = w * x + b
z.retain_grad()
```

```python
# 4. 定义损失函数计算损失
loss_fn = nn.MSELoss()
loss = loss_fn(z, y)
loss.retain_grad()
```

```python
# 5. 反向传播计算梯度: backward 必须是在标量上调用
loss.backward()
```

```python
print(loss.grad)
print(z.grad)
print(w.grad)
print(b.grad)
print(x.grad)
print(y.grad)
```

```python
z = w * x + b
z.retain_grad()
loss = loss_fn(z, y)
loss.backward()

print(loss.grad)
print(z.grad)
print(w.grad)
print(b.grad)
print(x.grad)
print(y.grad)
```

```python
print(x.is_leaf)
print(y.is_leaf)
print(w.is_leaf)
print(b.is_leaf)
print(z.is_leaf)
print(loss.is_leaf)
```

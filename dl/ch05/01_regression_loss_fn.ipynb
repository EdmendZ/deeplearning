```python
import torch
import torch.nn as nn
from fontTools.misc.py23 import tounicode
```

```python
## 构造数据 output神经网络的输出
output = torch.randn(5)
target = torch.randn(5)
print(output)
print(target)
```

## MAE(L1 Loss)

```python
loss_fn = nn.L1Loss()
loss = loss_fn(output, target)
print(loss)
```

## MSE(L2 Loss)

```python
loss_fn = nn.MSELoss()
loss = loss_fn(output, target)
print(loss)
```

## Smooth L1

```python
loss_fn = nn.SmoothL1Loss()
loss = loss_fn(output, target)
print(loss)
```

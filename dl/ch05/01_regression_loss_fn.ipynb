```python
import torch
import torch.nn as nn
# 回归损失函数：衡量「预测连续值」与「真实值」之间的数值差距
from fontTools.misc.py23 import tounicode
```

```python
# 构造数据：output 模拟神经网络的输出（预测值），target 为真实目标值
output = torch.randn(5)
target = torch.randn(5)
print(output)
print(target)
```

## MAE(L1 Loss)

```python
# MAE / L1：平均绝对误差 mean(|output - target|)，对异常值不敏感
loss_fn = nn.L1Loss()
loss = loss_fn(output, target)
print(loss)
```

## MSE(L2 Loss)

```python
# MSE / L2：均方误差 mean((output - target)^2)，对大误差惩罚更重、对异常值敏感
loss_fn = nn.MSELoss()
loss = loss_fn(output, target)
print(loss)
```

## Smooth L1

```python
# Smooth L1：误差小时像 MSE（平滑），误差大时像 L1（梯度不爆炸），兼顾两者
loss_fn = nn.SmoothL1Loss()
loss = loss_fn(output, target)
print(loss)
```
